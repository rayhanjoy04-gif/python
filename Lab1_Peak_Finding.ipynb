## Part 5. Measuring running time (slides 25–26)

We use the **worst case** for the linear scan: a strictly increasing array, where the only peak is the last element.

The linear scan is timed once. The binary version is so fast that one run is too short to measure, so we average over 1,000 runs.

*Expect the last row (n = 10⁷) to take a few seconds for the linear scan.*

````python
import time

sizes = [10**3, 10**4, 10**5, 10**6, 10**7]
results = []
for n in sizes:
    a = list(range(n))                      # worst case for the linear scan

    start = time.perf_counter()
    peak_linear(a)
    t_lin = time.perf_counter() - start

    start = time.perf_counter()
    for _ in range(1000):
        peak_binary(a)
    t_bin = (time.perf_counter() - start) / 1000

    results.append((n, t_lin, t_bin))
    print(f"n = {n:>10,}   linear = {t_lin*1000:10.3f} ms   binary = {t_bin*1e6:7.2f} µs   ratio = {t_lin/t_bin:12,.0f}x")
````
